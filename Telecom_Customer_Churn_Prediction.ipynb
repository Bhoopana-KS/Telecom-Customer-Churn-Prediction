```python
X = data_clean.drop("Churn",axis=1)
Y = data_clean["Churn"]
print(X.shape)
Y.shape
```

```python
numerical_cols = X.select_dtypes(include=["int64","float64"]).columns
numerical_cols
```

```python
categorical_cols = X.select_dtypes(include =["object","str","category"]).columns
categorical_cols
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42,stratify=Y)
```

```python
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", Y_train.shape)
print("Testing target:", Y_test.shape)
```

```python
print("Training Churn distribution:")
print(Y_train.value_counts(normalize=True) * 100)

print("\nTesting Churn distribution:")
print(Y_test.value_counts(normalize=True) * 100)
```

```python
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer(transformers=[("num",StandardScaler(),numerical_cols),
                                               ("cat",OneHotEncoder(handle_unknown="ignore"),categorical_cols)])
```

```python
X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)
```

```python
print("Original training shape:", X_train.shape)
print("Encoded training shape:", X_train_encoded.shape)

print("\nOriginal testing shape:", X_test.shape)
print("Encoded testing shape:", X_test_encoded.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
model = LogisticRegression(random_state=42,max_iter=1000)
```

```python
model.fit(X_train_encoded,Y_train)
```

```python
Y_pred = model.predict(X_test_encoded)
print("first 10 predictions:")
print(Y_pred[:10])
```

```python
accuracy = accuracy_score(Y_test,Y_pred)
print("Model accuracy:",accuracy*100,"%")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(Y_test, Y_pred, labels=["No", "Yes"])
plt.figure(figsize=(6, 4))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No", "Yes"],
    yticklabels=["No", "Yes"]
)
plt.xlabel("Predicted Churn")
plt.ylabel("Actual Churn")
plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

```python
print(classification_report(Y_test, Y_pred))
```

```python
print(cm)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)
rf_model.fit(X_train_encoded, Y_train)
rf_pred = rf_model.predict(X_test_encoded)
print("First 10 predictions:")
print(rf_pred[:10])
print(classification_report(Y_test, rf_pred))
rf_cm = confusion_matrix(
    Y_test,
    rf_pred,
    labels=["No", "Yes"]
)

plt.figure(figsize=(6, 4))

sns.heatmap(
    rf_cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["No", "Yes"],
    yticklabels=["No", "Yes"]
)

plt.xlabel("Predicted Churn")
plt.ylabel("Actual Churn")
plt.title("Confusion Matrix - Random Forest")

plt.show()
```

```python
print("LOGISTIC REGRESSION")
print(classification_report(Y_test, Y_pred))

print("\nRANDOM FOREST")
print(classification_report(Y_test, rf_pred))
```

```python
# Get probability of Churn = Yes
lr_prob = model.predict_proba(X_test_encoded)[:, 1]

# Predict using a threshold of 0.4
lr_pred_04 = (lr_prob >= 0.4).astype(int)

# Convert numerical predictions back to Yes and No
lr_pred_04 = ["Yes" if p == 1 else "No" for p in lr_pred_04]

# Evaluate the model
print(classification_report(Y_test, lr_pred_04))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.3, 0.4, 0.5]

for threshold in thresholds:

    # Convert probabilities into predictions
    predictions = [
        "Yes" if probability >= threshold else "No"
        for probability in lr_prob
    ]

    # Calculate evaluation metrics
    precision = precision_score(Y_test, predictions, pos_label="Yes")
    recall = recall_score(Y_test, predictions, pos_label="Yes")
    f1 = f1_score(Y_test, predictions, pos_label="Yes")

    print(f"\nThreshold: {threshold}")
    print("Precision:", round(precision, 3))
    print("Recall:", round(recall, 3))
    print("F1-score:", round(f1, 3))
```

```python
lr_pred_03 = [
    "Yes" if probability >= 0.3 else "No"
    for probability in lr_prob
]
lr_accuracy_03 = accuracy_score(Y_test, lr_pred_03)

print("Accuracy at threshold 0.3:",
      lr_accuracy_03 * 100)
lr_cm_03 = confusion_matrix(
    Y_test,
    lr_pred_03,
    labels=["No", "Yes"]
)

plt.figure(figsize=(6, 4))

sns.heatmap(
    lr_cm_03,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No", "Yes"],
    yticklabels=["No", "Yes"]
)

plt.xlabel("Predicted Churn")
plt.ylabel("Actual Churn")
plt.title("Logistic Regression - Threshold 0.3")

plt.show()
```

```python
final_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression (0.5)",
        "Logistic Regression (0.3)",
        "Random Forest (0.5)"
    ],

    "Accuracy": [
        accuracy_score(Y_test, Y_pred),
        accuracy_score(Y_test, lr_pred_03),
        accuracy_score(Y_test, rf_pred)
    ],

    "Precision": [
        precision_score(Y_test, Y_pred, pos_label="Yes"),
        precision_score(Y_test, lr_pred_03, pos_label="Yes"),
        precision_score(Y_test, rf_pred, pos_label="Yes")
    ],

    "Recall": [
        recall_score(Y_test, Y_pred, pos_label="Yes"),
        recall_score(Y_test, lr_pred_03, pos_label="Yes"),
        recall_score(Y_test, rf_pred, pos_label="Yes")
    ],

    "F1 Score": [
        f1_score(Y_test, Y_pred, pos_label="Yes"),
        f1_score(Y_test, lr_pred_03, pos_label="Yes"),
        f1_score(Y_test, rf_pred, pos_label="Yes")
    ]
})

print(final_comparison.round(3))
```

```python
import pandas as pd

# Get the importance score of each feature
feature_importance = rf_model.feature_importances_

# Get the names of the encoded features
feature_names = preprocessor.get_feature_names_out()

# Create a DataFrame
importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": feature_importance
})

# Sort features from highest to lowest importance
importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

# Display the top 10 important features
importance_df.head(10)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select the top 10 features
top_10_features = importance_df.head(10)

# Create the bar chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_10_features,
    x="Importance",
    y="Feature"
)

plt.title("Top 10 Features Influencing Customer Churn")
plt.xlabel("Feature Importance")
plt.ylabel("Features")

plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Create a pipeline containing preprocessing and the model
final_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        random_state=42,
        max_iter=1000
    ))
])

# Train the complete pipeline using the original training data
final_model.fit(X_train, Y_train)
```

```python
import joblib

# Store the model and prediction threshold together
model_bundle = {
    "model": final_model,
    "threshold": 0.3
}

# Save the bundle as a file
joblib.dump(model_bundle, "telecom_churn_model.pkl")

print("Model saved successfully!")
```
